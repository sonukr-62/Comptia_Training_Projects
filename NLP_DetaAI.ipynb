```python
text = "There were once Three Little Pigs! They lived in a straw hut, a stick hut, and a brick hut."

# lower case
normalized_text = text.lower()

# Remove punctuation
normalized_text = "".join([char for char in normalized_text if char.isalnum() or char.isspace()])

# Remove extra spaces
normalized_text = " ".join(normalized_text.split())

print("Original Text:", text)
print("Normalized Text:", normalized_text)
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
sentence = "Hello world."

tokens = nltk.word_tokenize(sentence)

print(tokens)
```

```python
# Sentence to tokenize
text = "This is a sample sentence for tokenization demo."

# Word Tokenization (splitting by whitespace)
word_tokens = text.split()
print("Word Tokens:", word_tokens)
# Output: ['This', 'is', 'a', 'sample', 'sentence', 'for', 'tokenization', 'demo.']

# Character Tokenization (splitting by character)
char_tokens = list(text)
print("Character Tokens:", char_tokens)
```

```python
# Stop words removal demo code:
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

# removing stop words
import re
text = input("Enter any text: ")
words = re.sub(r'[^\w\s]', '', text).split()  # Remove punctuation
filtered_words = [word for word in words if word not in stop_words]
```

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download necessary NLTK data
nltk.download('punkt')
nltk.download('wordnet')

text = "The cats were running and studying"

# Stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in text.split()]
print(stemmed_words)

# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in text.split()]
print(lemmatized_words)
```
